 a turma do 2 ano encontrou um envelope cheio de figurinhas. para garantir uma divisão justa , todo concordaram com a sugesstão de joão, o dono do envelope :
  caa amigo, exceto joão , deve receber exatamente o mesmo número de figurinhas. joão deve receber o dobro d quantidade que cada amigo recebe. pode ser que o ao  joão ser o dono do envelope tenha ajudado  convecer o amigo, ma também contribuiu o fato de que, do jeito proposto,a divisção era exata, sem sobrar nem alar figurinhas.
  problema:
  crie um program que determine a quantidade de figurinhas que joão recebeu, dado o número total de figurinhas no envelope e o número de amigo de joão.
  entrada:
    . primeira linha contém inteiro total_figurinhas que joão recebeu, dado o número total de figurinhas no envelope.
    . a segunda linha contém um número inteiro numero_amigos, que representa quantidade e amigo e joão (excluindo o próprio joão).

```
total_figurinhas = int (input("digite o total de figurinhas:"))
numero_amigo = int (input("digite o número de amigos:"))

figurinha_amigos = total_figurinhas // (numero_amigo + 2)
figurinhas_joao = 2 * figurinha_amigos

print(f"joão recebeu {figurinhas_joao} figurinhas.")
```
